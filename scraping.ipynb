## Translation

```python
translator = Translator()


async def translate(text, src, dest="en"):
    if not isinstance(text, str) or not text.strip():
        return text

    result = await translator.translate(text, src=src, dest=dest)
    return result.text


async def translate_texts(texts, src):
    tasks = [asyncio.create_task(translate(text, src)) for text in texts]
    translated_texts = await asyncio.gather(*tasks)
    return translated_texts


# Function to wrap async translation for pandas column.
async def translate_column_async(df, column_name, src_lang):
    texts = df[column_name].tolist()
    translated_texts = await translate_texts(texts, src_lang)
    df[f"en_{column_name}"] = translated_texts
    return df
```

```python
vi_df = pd.read_csv("data/app_reviews_vi.csv")

vi_df = await translate_column_async(vi_df, "content", "vi")
```

```python
id_df = pd.read_csv("data/app_reviews_id.csv")

id_df = await translate_column_async(id_df, "content", "id")
```

```python
ar_df = pd.read_csv("data/app_reviews_ar.csv")

ar_df = await translate_column_async(ar_df, "content", "ar")
```

## Storing
